```python
from sklearn.datasets import load_iris
import pandas as pd

# Load the dataset
iris = load_iris()
dir(iris)
```

```python
feature=iris.data
feature
```

```python
category = iris.target
category
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(feature,category,test_size=0.3)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
predicted_class = model.predict([iris.data[92]])[0]
flower_name = iris.target_names[predicted_class]
print("Predicted flower:", flower_name)
```
